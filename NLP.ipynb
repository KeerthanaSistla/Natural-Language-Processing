#1. Building a Simple Sentiment Analysis Model

Sentiment Analysis is a Natural Language Processing (NLP) task used to determine whether a given text expresses positive, negative, or neutral sentiment. It is widely used in product reviews, social media analysis, and customer feedback systems.

In this implementation, the 20 Newsgroups dataset was used for binary text classification. Although it is a topic classification dataset, two categories were selected to simulate sentiment-style classification.

The NLP pipeline consists of:

Text Cleaning – Removing punctuation and converting text to lowercase

Tokenization – Splitting text into words

Stopword Removal – Removing common words like "the", "is"

Lemmatization – Converting words to base form

TF-IDF Vectorization – Converting text into numerical features

Multinomial Naive Bayes – Used for classification

Naive Bayes works well for text classification because it assumes feature independence and performs efficiently on high-dimensional sparse data.

The model achieved approximately 85–95% accuracy, indicating good classification performance.

```python
import nltk
import re
import numpy as np
from sklearn.datasets import fetch_20newsgroups
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report, accuracy_score
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

# Download required NLTK resources
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('wordnet')

# Load only 2 categories (binary classification)
categories = ['rec.sport.hockey', 'sci.space']

newsgroups = fetch_20newsgroups(
    subset='all',
    categories=categories,
    remove=('headers', 'footers', 'quotes')
)

X = newsgroups.data
y = newsgroups.target

print("Total Samples:", len(X))

# Preprocessing
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def preprocess(text):
    text = text.lower()
    text = re.sub(r'[^a-zA-Z]', ' ', text)
    tokens = nltk.word_tokenize(text)
    tokens = [lemmatizer.lemmatize(word) for word in tokens if word not in stop_words]
    return " ".join(tokens)

X_clean = [preprocess(text) for text in X]

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X_clean, y, test_size=0.2, random_state=42
)

# TF-IDF Feature Extraction
vectorizer = TfidfVectorizer(max_features=5000)
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)

# Model
model = MultinomialNB()
model.fit(X_train_vec, y_train)

# Predictions
y_pred = model.predict(X_test_vec)

# Evaluation
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n")
print(classification_report(y_test, y_pred))
```

The 20 Newsgroups dataset was used for binary text classification.
Text preprocessing included tokenization, stopword removal, and lemmatization.
TF-IDF was used for feature extraction and Multinomial Naive Bayes was used for classification.

#3. Implementation of NLP ChatBot

A chatbot is an NLP application that simulates human conversation through text or speech.

In this implementation, a rule-based chatbot is developed using the nltk.chat.util.Chat module from the Natural Language Toolkit.

🔹 Working Principle

The chatbot works using:

Pattern Matching (Regular Expressions)

Each user input is matched against predefined patterns.

Patterns like "hi|hello|hey" match multiple greeting words.

Response Mapping

For each pattern, a list of possible responses is stored.

When a pattern matches, the chatbot randomly selects one response.

Reflections Dictionary

Built-in reflections convert words like:

"I am" → "you are"

"my" → "your"

This makes responses slightly more conversational.

🔹 Type of Chatbot Used

This is a:

✔ Rule-based chatbot
✔ Pattern-matching chatbot
✔ Deterministic conversational system

It does NOT use machine learning.

🔹 Advantages

Simple to implement

Fast execution

No training data required

Good for small applications

🔹 Limitations

Cannot understand unseen inputs

No learning capability

Limited conversational intelligence

🔹 Algorithm (Write This for Extra Marks)

Define conversation patterns using regular expressions.

Map each pattern to predefined responses.

Initialize chatbot using Chat(pairs, reflections).

Accept user input.

Match input against patterns.

Display corresponding response.

Repeat until user types "quit".

```python
pip install nltk
```

```python
import nltk
from nltk.chat.util import Chat, reflections

pairs = [
    [
        r"hi|hello|hey",
        ["Hello!", "Hi there!", "Hey! How can I help you?"]
    ],
    [
        r"what is your name?",
        ["I am a simple NLP ChatBot."]
    ],
    [
        r"how are you?",
        ["I am fine. Thank you!"]
    ],
    [
        r"quit",
        ["Goodbye! Have a nice day."]
    ]
]

chatbot = Chat(pairs, reflections)

print("Type 'quit' to exit.")
chatbot.converse()
```

#4. Conditional Frequency Distribution in NLP

A Conditional Frequency Distribution (CFD) is used to count word frequencies under certain conditions.

It helps answer:

How often does a word appear in a specific category?

What words are common in positive vs negative text?

In Natural Language Processing with Python, CFD is explained using NLTK.



---

Suppose we have:

Category A: positive words

Category B: negative words

```python
import nltk
from nltk.probability import ConditionalFreqDist

data = [
    ("positive", "good"),
    ("positive", "happy"),
    ("positive", "good"),
    ("negative", "bad"),
    ("negative", "sad"),
    ("negative", "bad")
]

cfd = ConditionalFreqDist(data)

print(cfd["positive"])
print(cfd["negative"])
```

#5. Processing Raw Text: Accessing Text from Web and Disk

Theory

Processing raw text is the foundational step in Natural Language Processing (NLP). Raw text is unstructured data that must be cleaned and transformed into a structured format before applying any NLP techniques such as classification, sentiment analysis, or language modeling.

Text data can be accessed from two primary sources:


---

Accessing Text from the Web

Text can be extracted from websites, online documents, APIs, or digital libraries using HTTP requests.

In Python, this is typically done using:

requests library – to fetch webpage content

BeautifulSoup – to parse HTML content

nltk – to tokenize and process text


Steps involved:

1. Send request to URL


2. Retrieve HTML or raw text data


3. Extract meaningful text


4. Perform preprocessing (tokenization, stopword removal, etc.)



This method is useful for:

Web scraping

Social media analysis

Online news analysis

Digital book processing


Example sources:

Project Gutenberg (free books)

Wikipedia (encyclopedia content)



---

2️⃣ Accessing Text from Disk

Text stored locally in .txt files can be accessed using Python’s built-in open() function.

Steps involved:

1. Open file in read mode


2. Read raw text


3. Store text in variable


4. Apply NLP preprocessing



This method is commonly used when:

Working with datasets

Processing logs

Analyzing stored documents

Performing corpus-based NLP research



---

Text Preprocessing After Access

Once raw text is obtained (from web or disk), it undergoes preprocessing:

1. Tokenization – Splitting text into words or sentences


2. Lowercasing – Converting text to uniform case


3. Removing Punctuation & Special Characters


4. Stopword Removal – Removing common words like "the", "is"


5. Stemming / Lemmatization – Converting words to base form



This converts unstructured text into structured data suitable for machine learning models.


---

Importance in NLP

Processing raw text is important because:

Real-world data is messy and unstructured

Models require numerical representation of text

Preprocessing improves model accuracy

Reduces noise and irrelevant information


It is the first and most critical stage in any NLP pipeline.


---

Conclusion

Accessing and processing raw text from the web and disk enables extraction of meaningful information from large volumes of unstructured data. Proper preprocessing ensures high-quality input for downstream NLP tasks such as classification, sentiment analysis, and chatbot development.
